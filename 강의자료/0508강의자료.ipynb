## 실습: 점수 분석기

## 클래스 활용 예-1
*  선박의 배수량(Displacement)과 복원성(Stability) 지표 중 하나인 GM(Metacentric Height)을 계산하는 간단한 클래스 예제 프로그램

````python
class Ship:
    # [클래스 변수] 모든 선박이 공유하는 물리적 상수 (예: 바닷물 밀도)
    SEA_WATER_DENSITY = 1.025  # ton/m^3

    def __init__(self, name, length, beam, draft, cb):
        """
        [생성자] 선박의 초기 제원을 입력받아 멤버 필드를 선언합니다.
        """
        self.name = name      # 선박 이름
        self.length = length  # 길이 (LBP, m)
        self.beam = beam      # 폭 (m)
        self.draft = draft    # 흘수 (m)
        self.cb = cb          # 방형계수 (Block Coefficient)

        # 기본값으로 설정할 멤버 필드
        self.kg = 0.0         # 중심 높이 (m)
        self.km = 0.0         # 메타센터 높이 (m)

    def calculate_displacement(self):
        """
        [메소드] 배수량을 계산합니다. (L * B * d * Cb * 밀도)
        """
        displacement = self.length * self.beam * self.draft * self.cb * self.SEA_WATER_DENSITY
        return round(displacement, 2)

    def set_stability_params(self, kg, km):
        """
        [메소드] 복원성 계산을 위한 KG와 KM 값을 설정합니다.
        """
        self.kg = kg
        self.km = km

    def get_gm(self):
        """
        [메소드] GM(메타센터 높이)을 계산하여 복원성 판단을 돕습니다.
        """
        gm = self.km - self.kg
        status = "양호" if gm > 0 else "위험 (복원력 상실)"
        return gm, status

# --- 프로그램 실행 부분 ---

# 1. 인스턴스 생성 (컨테이너선 예시)
my_ship = Ship(name="BlueWhale", length=300, beam=45, draft=12, cb=0.7)

# 2. 멤버 필드 접근 및 출력
print(f"선박명: {my_ship.name}")
print(f"계산된 배수량: {my_ship.calculate_displacement()} tons")

# 3. 메소드를 통한 상태 업데이트 (복원성 데이터 입력)
my_ship.set_stability_params(kg=15.5, km=18.2)

# 4. 최종 결과 출력
gm_value, safety = my_ship.get_gm()
print(f"현재 GM: {gm_value:.2f}m -> 상태: {safety}")
````

## 클래스 활용 예-2
* 여러척의 선박 데이터 관리 프로그램 (클래스와 리스트를 활용)

````python
class Ship:
    SEA_WATER_DENSITY = 1.025

    def __init__(self, name, length, beam, draft, cb):
        self.name = name
        self.length = length
        self.beam = beam
        self.draft = draft
        self.cb = cb

    def calculate_displacement(self):
        displacement = self.length * self.beam * self.draft * self.cb * self.SEA_WATER_DENSITY
        return round(displacement, 2)

    def get_info(self):
        """선박의 주요 정보를 문자열로 반환하는 메소드"""
        return f"[{self.name}] L:{self.length}m, B:{self.beam}m, d:{self.draft}m, Cb:{self.cb}"

# --- 리스트를 활용한 선박 관리 ---

# 1. Ship 클래스의 인스턴스들을 원소로 가지는 리스트 생성
fleet = [
    Ship("Container-A", 300, 45, 12.5, 0.70),
    Ship("Tanker-B", 330, 60, 20.0, 0.82),
    Ship("LNG-C", 290, 46, 11.5, 0.75)
]

print("=== 선박 함대(Fleet) 정보 출력 ===")

# 2. 반복문을 사용하여 리스트 내의 각 객체 정보 출력
for ship in fleet:
    # 각 객체의 메소드 호출
    info = ship.get_info()
    disp = ship.calculate_displacement()

    print(f"{info} | 배수량: {disp} tons")

# 3. 특정 조건 검색 예시 (예: 배수량이 30만 톤 이상인 배 찾기)
print("\n=== 대형 선박 검색 (300,000 tons 이상) ===")
for ship in fleet:
    if ship.calculate_displacement() >= 300000:
        print(f"검색 결과: {ship.name} (배수량: {ship.calculate_displacement()} tons)")
````
