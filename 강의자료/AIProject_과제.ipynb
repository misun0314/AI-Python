# 실습1. AI 챗봇 만들기

* 사용자와 AI가 간단한 대화를 수행하는 프로그램만들어보기

* 요구사항 : 사용자가 종료 또는 exit 를 입력하면 대화종료

* 학습자가 알아야 할 내용
  * input()
  * while 반복문
  * 조건문
  * 문자열 처리
  * AI 프롬프트 작성

1. 직접 실행해보기

```python
while True:
  Q=input("질문입력 :")
  if Q in ("종료","exit") :
    print("AI : 종료합니다.")
    break
  print(f"AI : {Q}에 대한 답변입니다.")
```
