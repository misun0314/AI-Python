## 6. 자료 구조

* 리스트 : 순서가 있는 자료구조
  * 특징:
    - 여러 값을 한 번에 저장
    - 인덱스는 0부터 시작
    - 서로 다른 자료형 저장 가능
    - 수정 가능(mutable)



````python
nums = [10, 20, 30]
print(nums)
nums[1] = 99
print(nums)      # [10, 99, 30]
````

````python
# 사용자로부터 학생수를 입력받고, 학생점수 리스트 만들기
score=[]
print(type(score))
count=int(input("학생수입력:"))
for i in range(0,count):
  score.append(int(input("점수입력:")))
print(score)
````

````python
#grade 함수 사용하여 학생등급 리스트 만들기
std_grade=[]
for i in range(0,count):
  std_grade.append(grade(score[i]))
print(std_grade)
````

````python
total = 0

for s in score:
    total += s

avg = total / len(score)
print("평균:", avg)
````

###NumPy
* 기본 자료 구조 : 배열
* 같은 타입의 데이터를 배열로 관리
* 벡터화(원소별) 연산이 기본
* 다차원 배열인 ndarray(n-dimension array)
  * 메모리를 효율적으로 사용하여 연산속도가 빠름
  * 벡터 산술연산 기능 제공
  * 브로드캐스팅 기능 제공: shape이 다른 다차원 배열 또는 스칼라 값 사이에서 연산이 가능
* 반복문을 작성할 필요 없이 전체 데이터 배열에 대해 빠른 연산을 제공

````python
arr = np.array([1, 2, 3])
print("리스트 *2:", [1,2,3]*2)
print("NumPy 배열 *2:", arr*2)

m = np.arange(1,7).reshape(2,3)
print("shape:", m.shape)
print(m)
print("열별 최대:", m.max(axis=0))
print("행별 최대:", m.max(axis=1))
````

### 시리즈(Series)와 데이터프레임(DataFrame)
* 시리즈(Series)
  * 1차원 배열로서 단일 변수 데이터를 저장하는 데 사용된다.
* 데이터프레임(DataFrame)
  * 2차원 배열로서 다중 변수 데이터를 저장하는 데 사용된다.
  * 데이터프레임은 여러 개의 1차원 배열(시리즈)을 세로 방향으로 묶어 놓은 형태를 가진다.

````python
# Series 생성 ((1차원 데이터)레이블 및 이름 포함)
temperatures = pd.Series([23.5, 25.1, 19.8, 30.2],
                       index=['월', '화', '수', '목'],
                        name='온도')
print(temperatures)
````

````python
print(temperatures.iloc[2])
````

````python
print(temperatures.loc['수'])
````

````python
# DataFrame 생성 (2차원 데이터)
weather_data = pd.DataFrame({
    '온도': [23.5, 25.1, 19.8, 30.2],
    '습도': [60, 65, 70, 55],
    '강수량': [0, 0, 5.2, 0]
}, index=['월', '화', '수', '목'])

print(weather_data)
display(weather_data)
````

````python
# 한 행 선택 (예: '월'요일 데이터)
print(weather_data.loc['월'])
````

````python
# 한 열 선택 (예: '습도' 컬럼 데이터)
print(weather_data['습도'])
````
