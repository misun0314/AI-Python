# 4.5.1 실습 - 국민건강보험공단 건강검진정보 데이터

### 데이터 준비 및 전처리

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib import font_manager, rc
plt.rcParams["font.family"] = 'NanumGothic'
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/미래혁신인재양성과정/projectlab/국민건강보험공단_건강검진정보_2024.CSV", encoding = 'euc-kr')
```

### 데이터 정보 확인

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

### 불필요한 컬럼 제거 및 컬럼명 변경

```python
df.drop(['기준년도', '가입자일련번호', '결손치 유무', '치아마모증유무', '제3대구치(사랑니) 이상'], inplace=True, axis = 1)
```

```python
df.rename(columns = {'연령대코드(5세단위)':'연령코드', '신장(5cm단위)':'신장', '체중(5kg단위)': '체중', \
                     '식전혈당(공복혈당)': '혈당' }, inplace = True)
```

```python
df = df.dropna()
```

###  전체 데이터 분석 시각화

```python
fig, axs = plt.subplots(5, 5)
fig.set_size_inches(20,24)

for i in range(5):
    for j in range(5):
        attr = i * 5 + j + 1
        if df[df.columns[attr]].nunique() < 30:
            sns.countplot( x=df.columns[attr], data=df,ax=axs[i][j])
        else:
            sns.histplot( x=df.columns[attr], data=df, kde=True, ax=axs[i][j])
plt.show()
```

```python
df['음주여부'].value_counts()
```

### 혈압 데이터 상관관계

```python
sns.scatterplot(x=df['수축기혈압'], y=df['이완기혈압'], hue=df['흡연상태'])
plt.show()
```

### 신장과 체중 상관관계

```python
sns.scatterplot(x=df['신장'], y=df['체중'], hue=df['성별코드'])
plt.show()
```

### 혈당과 총 콜레스테롤 상관관계

```python
sns.scatterplot(x=df['혈당'], y=df['총콜레스테롤'], hue=df['성별코드'])
plt.show()
```

## 실습 1. 이상치 제거 후 재작성

### 나이에 따른 총 콜레스테롤 추이

```python
sns.lineplot(x=df['연령코드'], y=df['총 콜레스테롤'])
plt.show()
```

### 나이에 따른 혈색소 수치 분포

```python
fig = plt.figure(figsize=(10,5))
sns.boxplot(x=df['연령코드'], y=df['혈색소'])
plt.show()
```

### 연령과 성별에 따른 혈당 분석

```python
fig = plt.figure(figsize=(12,5))
sns.barplot(x=df['연령코드'], y=df['혈당'], hue=df['성별코드'])
plt.show()
```

### 나이에 따른 허리둘레 분포

```python
fig = plt.figure(figsize=(12,5))
sns.violinplot(x=df['연령코드'], y=df['허리둘레'], hue=df['성별코드'])
plt.show()
```

### 지역과 연령에 따른 혈당 수치 히트맵

```python
pivot_df = df.pivot_table("혈당", "시도코드", "연령코드")
sns.heatmap(pivot_df)
plt.show()
```
