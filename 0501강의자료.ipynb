1. 선박 제원 출력 및 배수량 계산문제: 사용자의 입력을 받아 선박의 이름, 길이(L), 폭(B), 흘수(T)를 변수에 저장하고, 단순화된 공식을 이용해 배수량($\Delta$)을 계산하여 출력하는 프로그램을 작성하세요.
* 공식: $Displacement = L \times B \times T \times C_b \times 1.025$ (단, 방형비 $C_b$는 0.8로 가정)

```python

ship_name = input("선박 이름을 입력하세요: ")
length = float(input("선박의 길이(L, m)를 입력하세요: "))
width = float(input("선박의 폭(B, m)를 입력하세요: "))
draft = float(input("선박의 흘수(T, m)를 입력하세요: "))
cb = 0.8  # 방형비

displacement = length * width * draft * cb * 1.025

print(f"\n=========={ship_name} 제원 정보 ===========")
print(f"길이: {length}m, 폭: {width}m, 흘수: {draft}m")
print(f"계산된 배수량: {displacement:.2f} tons")
print("="*33)
```

## 누적합/누적곱 함수로 만들어보기

```python
total = 0
fac = 1
for i in range(1, 6):
    total = total + i
    fac = fac * i
print("sum(1..5):", total)
print("fac(1..5):", fac)
```

```python
def sum(start, end) :
  total=0
  for i in range(start, end+1):
    total = total + i
  return total
def fac(start, end) :
  fac = 1
  for i in range(start, end+1):
    fac = fac * i
  return fac

num1=int(input("정수1 입력:"))
num2=int(input("정수2 입력:"))
print(f"sum({num1}..{num2}):", sum(num1,num2))
print(f"fac({num1}..{num2}):", fac(num1,num2))
```
